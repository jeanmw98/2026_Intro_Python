**Q1) Import geopandas and pandas.**

Import geopandas as `gpd`, pandas as `pd`, and matplotlib's pyplot as `plt`.

```python
# Import geopandas (as gpd), pandas, and matplotlib
import geopandas as gpd
import pandas as pd
import matplotlib as plt
```

**Q2) Read the state boundary file with geopandas' `read_file` function.**

The file is cached at `states_path`. Call the result `country`.

```python
# Read the data with the geopandas function
country = gpd.read_file(states_path)
```

**Q3) Have a look at the data. What type of geometries does it contain?**

Print the first few rows, and answer the geometry question in a comment.

```python
# Print out the first few lines of the data : Here i am printing the first 5 lines (o to 4) thanks to .head()
print(country.head(5))
```

**Q4) Have a look at the data on a map using geopandas' `.plot()` method.**

Exclude Alaska and Hawaii using the `NAME` attribute and pandas' `.isin()` method. Specify the
figsize to be 30 x 20.

```python
# Exclude Alaska and Hawaii : Here country_out is given by taking away the name of states from the country (~country["NAME"]) and flaging the states from our Series we want out (["Alaska", "Hawaii"]).
country_out = country[~country["NAME"].isin(["Alaska", "Hawaii"])]

# Plot the US states with figsize 30 x 20 : i plot country_out using (.plot) and giving it the required figsize (figsize=). The colour of USA is not chosen and has automaticaly been selected to blue.
country_out.plot(figsize=(30, 20))
```

```python
# Pre-supplied: read in the hurricane Florence data, drop the advisory bookkeeping columns,
# fix the longitude sign, and have a look at the dataframe.
# The Long column holds a positive "degrees west" magnitude, not a signed longitude -- a common
# quirk of NHC advisory data.
florence = pd.read_csv(track_path)
florence = florence.drop(["AdvisoryNumber", "Forecaster", "Received"], axis=1)
florence["Long"] = 0 - florence["Long"]
florence.head(3)
```

**Q5) Create a GeoDataFrame from the `florence` DataFrame.**

Build the geometry column from the `Long` and `Lat` columns with
[`gpd.points_from_xy`](https://geopandas.org/en/stable/docs/reference/api/geopandas.points_from_xy.html),
and call the result `gdf_florence`. Then look at its first few rows.

```python
# Create a geodataframe from the hurricane florence dataframe : Here i am modifying (florence) which is a pandas DataFrame csv file, to a GeoDataFrame file. It is easily done with (gpd.GeoDataFrame) and points_from_xy using the longitude and latitude of hurricane florence's points. Finally, i choose the correct EPSG coordinates for a world map.
#gdf_florence = gpd.GeoDataFrame(florence, geometry=gpd.points_from_xy(florence["lon"], florence["lat"]), crs="EPSG:4326") i simply didn't use the right names for longitude and latitude present in the Series
gdf_florence = gpd.GeoDataFrame(florence, geometry=gpd.points_from_xy(florence["Long"], florence["Lat"]), crs="EPSG:4326")
```

**Q6) Plot the US states map (without Alaska and Hawaii) and hurricane Florence together.**

Draw the states as a base map, then plot the hurricane positions on top in a colour that stands
out.

```python
# Plot to see the hurricane overlay the US map, with the hurricane position on top : Here i create ax that plot the same map of the USA using the same figsize, but changing the colour of the states to lightgray to increase visibility and adding a black edge to the states
# then i ask to plot florence with ax=ax which tells GeoPandas to draw the hurricane on the existing USA map, rather than creating a separate figure. I chose the colour red "r" and a thick markersize of 50 so we can see the hurricane well.
ax = country_out.plot(figsize=(30, 20),facecolor='lightgray', edgecolor='k')
gdf_florence.plot(ax=ax, color='r',markersize=50)
```

**Q7) What is the coordinate reference system of the data?**

Check it for both layers, and say in a comment whether they match — two layers only line up on
one pair of axes if they share a crs.

```python
# Check the coordinate reference system of the data : it is "EPSG:4326" for both geodataframes, which are world coordinates. (.crs) gives this information
print(country_out.crs)
print(gdf_florence.crs)
```

**Q8) Which states did the hurricane pass through?**

**ℹ️ Hint**

One approach is to plot the trajectory over the US map, annotate each state with its name, and
read the answer off the figure. Other approaches are more than welcome — `overlay` with
`how="intersection"` will give you the track points that fall inside a state polygon, and the
`NAME` column of the result then answers the question directly.

```python
import matplotlib.pyplot as plt

# I am creating clip that finds the states affected by hurricane florence thanks to (gpd.overlay()), with gdf_florence inside/"inner" country_out
#clip = gpd.overlay(gdf_florence, country_out, how="inner")
# I receiced a ValueError: `how` was 'inner' but is expected to be in ['intersection', 'union', 'identity', 'symmetric_difference', 'difference']
# This means that "inner" is not expected for .overlay, but i should rather use the enumerated values []
# Here is the corrected line using "intersection"
# Another way to do this is with the (.join) method we saw in class, in my error, i might have got mixed with this method as .join uses 'inner' as intercept
clip = gpd.overlay(gdf_florence, country_out, how="intersection")

# Here i print the names of the states that are affected by the hurricane using (.unique) with the ([NAME] of states inside clip)
# The states affected by the hurricane are shown before the map and are : 'North Carolina' 'South Carolina' 'Virginia' 'Kentucky' 'West Virginia'
#'Pennsylvania' 'Massachusetts'
print("States affected by hurricane florence :", clip["NAME"].unique())

# Again like in Q6, i plot the map of USA and the hurricane with the same settings, but clip only shows the hurricane within the country and not it's section passing through the ocean.
# I added the label "Inner trajectory" to show what the red dots represent on the map
ax = country_out.plot(figsize=(30, 20), color='lightgrey', edgecolor='k')
clip.plot(ax=ax, color='r', markersize=50, label='Inner trajectory')

# Plotting the legend : The next step of presentation i would like to learn is a way to cleanly present the legend by adding the north, a ruler, title, author name... just like in GIS softwares
plt.legend()
```
